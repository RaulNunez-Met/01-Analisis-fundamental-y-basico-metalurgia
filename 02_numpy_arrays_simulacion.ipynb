```python
# Se utilizará numpy para generar datos aleatorios de planta como pueden ser mediciones de acidez hechas al flujo de riego
import numpy as np

# Definir un numero de mediciones a generar
numero_de_mediciones = 5000 

"""
# Usualmente para la solución de riego se requiere una concentración de acidez que permita tener un ph entre los rangos 2.0 a 3.0
# Los datos simulados se generaran entre un rango de ph de 1.5 a 3.5 
# Un pH alto genera problemas en la disolución del mineral y puede generar precipitados no deseados como la jarosita a pH > 3.0
# Por otro lado un pH muy bajo es indeseable, ya que se desperdician reactivos y se puede disolver material indeseado
"""

# Generamos 5000 valores de ph entre 1.5 y 3.5
mediciones_acidez = np.random.uniform(low=1.5, high=3.5, size=numero_de_mediciones)

print(f"Total de datos generados: {mediciones_acidez.size}")
print(f"5 datos aleatorios: {np.random.choice(mediciones_acidez, size=5, replace=False)};")
```

```python
# Cálculos estadísticos
acidez_promedio = mediciones_acidez.mean()
acidez_variabilidad = mediciones_acidez.std()

print(f"--- Análisis de Acidez --- ")
print(f"- Acidez promedio de las muestras: {acidez_promedio:.2f} g/L")

# Desviación estándar como medida de variabilidad, si es alta indica fallos en la preparación de la solución de riego
print(f"- Variabilidad (estabilidad): {acidez_variabilidad:.3f}")
print("\n--- Correción de mediciones y detección de anomalías ---")

# OPERACIÓN MASIVA: Poner al prueba el numpy corrigiendo todas las mediciones a la vez, simulando una corrección de algun medidor defectuoso en planta.
acidez_corregida = mediciones_acidez + 0.05

# Encontrar todos los días con anomalías
dias_acidez_baja = acidez_corregida[acidez_corregida > 3.0] 
dias_acidez_alta = acidez_corregida[acidez_corregida < 2.0]
print(f"- Total de días con pH > 3.0 en la soluciónd de riego (pH alto): {dias_acidez_baja.size} días.")
print(f"- Total de días con pH < 2.0 en la soluciónd de riego (pH bajo): {dias_acidez_alta.size} días.")
```
