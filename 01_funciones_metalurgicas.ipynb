```python
def calcular_eficiencia_extraccion(cobre_extraido: float, cobre_total_en_mineral: float) -> float:

    """
    cobre_extraido=cantidad de cobre extraído (en toneladas)
    cobre_total_en_mineral=ley del mineral (float para condicionar decimal, por ejemplo, 1.0 para 1%) 
    """
    # Corrección de errores (evitar división por cero y datos negativos)
    if cobre_total_en_mineral <= 0:
        print("¡ADVERTENCIA! Ley de cobre en mineral es cero o negativa.")
        return 0.0
    
    # Cálculo de la eficiencia de extracción
    eficiencia = (cobre_extraido / cobre_total_en_mineral) * 100.0
    
    # Evaluación de resultados
    if eficiencia > 90.0:
        print(f"Eficiencia Excelente: {eficiencia:.2f}% ¡Objetivo superado!")
    elif eficiencia > 75.0:
        print(f"Eficiencia Aceptable: {eficiencia:.2f}%. Dentro del rango meta.")
    else:
        print(f"¡ALERTA! Eficiencia Baja: {eficiencia:.2f}%. Revisar pH o flujo de riego.")
        
    return eficiencia

# --- PRUEBAS ---
# Caso 1: Extracción exitosa cobre_extraido=0.9, cobre_total_en_mineral=1.0
print("\n CASO 1:")
resultado_a = calcular_eficiencia_extraccion(cobre_extraido=0.9, cobre_total_en_mineral=1.0)

# Caso 2: Bajo desempeño cobre_extraido=0.6, cobre_total_en_mineral=1.0
print("\n CASO 2:")
resultado_b = calcular_eficiencia_extraccion(cobre_extraido=0.6, cobre_total_en_mineral=1.0)
```

```python
def monitorear_consumo_acido(consumos_diarios: list):  
    
# consumos_diarios: lista de consumos diarios de ácido en kg/tonelada
    consumo_total = 0
    dias_anomalos = 0
    dias = 0
    for consumo in consumos_diarios:
        consumo_total += consumo
        dias += 1
        # Detector de anomalías 
        if consumo>35.0 or consumo<20.0:
            print(f"Advertencia: Consumo de {consumo} kg/t el día {dias} (ANOMALÍA).")
            dias_anomalos += 1

    consumo_promedio= consumo_total/len(consumos_diarios)
    print("--------RESUMEN DEL MONITOREO--------")
    print(f"Dias totales: {len(consumos_diarios)}")
    print(f"Días con anomalías: {dias_anomalos}")
    print(f"El consumo total de ácido fue de: {consumo_total:.2f} kg/tonelada")
    print(f"El consumo promedio de ácido fue de: {consumo_promedio:.2f} kg/tonelada")

    
# --- DATOS DE PRUEBA ---
consumos = [16, 26.1, 34.9, 24.8, 41.0, 27.5, 25.0]
monitorear_consumo_acido(consumos)
```
